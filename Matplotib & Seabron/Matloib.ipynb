```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 5, 11)
x ## ye x axis ka value hai
y = x**2
y ## ye y axis ka value
```

```python
plt.plot(x, y) ## yese graph show hoga
plt.title("My first plot") # ye top me show hoga hoga graph ka
plt.xlabel("x axis") # ye x-axis me show hoga hoga msg
plt.ylabel("y axis") # ye y-axis me show hoga hoga msg
```

```python
##Subplots
plt.subplot(2,2,1) ## (rows, columns, in a graph area selected the coordinted 1,23,4)
plt.subplot(2,2,4) # selected the fourth cordinated to draw a graph
plt.plot(x, y) ## yese graph show hoga 4th cordinated wala
plt.subplot(2,2,3)
plt.plot(y,x) ## yese graph show hoga 3th cordinated wala

plt.subplot(2,2,2)
plt.plot(x**2, x**2) ## yese graph show hoga 4th cordinated wala
```

```python
## Object oriented way
fig = plt.figure(figsize=(8,12),dpi=100) ## (width, height) are set the figure size
axis1 = fig.add_axes([0.1, 0.1, 1, 1]) ##(left, bottom, width, height) isko kaha se show hoga
axis1.plot(x, y)
axis2 = fig.add_axes([0.3, 0.4, .3, .3]) # yaha par ham ek image ke ander othr image create nkiya hai
axis2.plot(y, x)
## yaha par value pass kar ke ham isko shifted kar rahi hai

```

```python
##Types of plots
plt.scatter(x, y) ## yese graph ... kar ke graph show hoga
```

```python
plt.hist(x) ## yese graph doumble kar ke graph show hoga
```

```python
plt.boxplot(x) ## yese graph doumble kar ke graph show hoga 
```

```python
## image ko ham save kar sakte hai
fig.savefig("myplot.png")
```
