```python
import matplotlib.pyplot as plt





import numpy as np








import seaborn as sns





import plotly.graph_objects as go





import plotly.express as px





import cufflinks as cf








cf.go_offline()
# Load the tips dataset








tips = sns.load_dataset('tips')








tips.head()
# Create interactive plot using cufflinks








tips['total_bill'].iplot(kind='hist', title='Distribution of Total Bill')
# Alternative using Plotly Express








fig = px.histogram(tips, x='total_bill', title='Distribution of Total Bill')






fig.show()
tips.iplot ( kind = 'scatter', y="tip" )





```
