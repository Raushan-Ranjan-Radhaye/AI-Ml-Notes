```python
import numpy as np
import pandas as pd
```

```python
data = {
    'Category': ['A', 'A', 'A', 'B', 'B', 'B'],
    "Store": ["A1", "A2", "A3", "B1", "B2", "B3"],
    "Sales":[100, 200, 300, 400, 500, 600],
    "Quantity": [10, 20, 30, 40, 50, 60],
    "data":pd.date_range("2022-01-01", periods=6)
}

df = pd.DataFrame(data)
```

```python
df['Sales'].describe()
```

```python
##Group by Category and Calcaulate the sum of sales and quantity

cat = df.groupby("Category")['Sales'].sum()
cat



```

```python
## Group by multiple columns
##Group byu Category and Store

cat = df.groupby(["Category", "Store"])['Sales'].sum()
##catergory aur Store dono ko sum kaerag
cat
```
