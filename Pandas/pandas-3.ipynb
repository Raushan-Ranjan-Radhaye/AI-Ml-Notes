```python
import numpy as np
import pandas as pd
```

```python
data = {
    'A': [1,2,np.nan,4,5],## nan ka matlab hai ki wo null value hai
    'B':[1,2,3,4,5],
    'C':[1,2,3,np.nan,np.nan],
    'D':[1,np.nan,np.nan, np.nan,5]
}

df = pd.DataFrame(data)


```

```python
df
```

```python
df.isna() ## used to print the null value and actual values in a from of true or false
```

```python
df.isna().sum() ## used to count the null value
```

```python
df.isna().any()## deaga ki kon kon sa me null value hai hai ya nahi to true or false return karega
```

```python
df.dropna()## yaha par ham null value ko delete karenge
```

```python
df.dropna(thresh=3) ## ye check karega ki data me kam se kam three null value hai ya nahi hoga to usko remove karega
```

```python
df.fillna(0) ## yaha par ham null value ko 0 se replace karenge
```

```python
values = {'A': 0, 'B':100, 'C':300, 'D':400}
df.fillna(values)## tah par ham column wise data ko fill kar rahi hai
```

```python
df.fillna(df.mean())
```
