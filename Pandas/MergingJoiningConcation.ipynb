```python
import numpy as np
import pandas as pd
```

```python
## merging two dataframes
employees = pd.DataFrame({
    'employee_id': [1, 2, 3, 4],
    'salary': [10000, 20000, 30000, 40000],
    'bonus': [5000, 6000, 7000, 8000]
})


salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 4],
    'salary': [10000, 20000, 30000, 40000],
    'bonus': [5000, 6000, 7000, 8000]
})
```

```python
employees
```

```python
print(salaries)
```

```python
pd.merge(employees, salaries, on='employee_id', how='inner')
## ham yaha par ham 2 dataframe ko merge kar rahi hai aur employee_id ke basices par ham
##ander me jo employee_id hai uske ko common id hai uske basis par merage karega dataframe ko
```

```python
pd.merge(employees, salaries, on='employee_id', how='outer')
```

```python
##Contation of two dataframe
df1 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2', 'A3'],
    'B':['B0', 'B1', 'B2', 'B3'],
    'C': ['C0', 'C1', 'C2', 'C3'],
})

df2 = pd.DataFrame({
    'A': ['A3', 'A4', 'A5', 'A6'],
    'B': ['B3', 'B4', 'B5', 'B6'],
    'C': ['C3', 'C4', 'C5', 'C6'],
})

```

```python
df1
```

```python
df2
```

```python
pd.concat([df1, df2])
```

```python
pd.concat([df1, df2], axis=1) ##we are concat the basis of row
```

```python
##Joining of two dataframe

##first dataframe
df1 = pd.DataFrame({
    'name':['Alice', 'Bob', 'Charlie'],
}, index=[1, 2, 3])

##second dataframe
df2 = pd.DataFrame({
    'score': [85, 92, 78]
}, index=[2,3,4])
```
