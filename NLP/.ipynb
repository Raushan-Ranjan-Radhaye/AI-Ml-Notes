```python
df['text'] = df['text'].apply(lambda x: x.lower())
# ye text column me jo text hai usko lowercase me convert kar raha hai
# lambda function is used to apply the function to each element of the text column
# x is the input to the lambda function and x.lower() is the output of the lambda function
# df['text'] is the text column of the dataframe and .apply() is used
```

```python
import string
def remove_Punc(text):
    return text.translate(str.maketrans('', '', string.punctuation))
# ye remove_Punc function me text ko input le raha hai aur usme se punctuation ko remove kar raha hai
# string.punctuation is a string that contains all the punctuation characters
# str.maketrans() is a method that creates a translation table that can be used to replace or remove characters
# in this case we are replacing punctuation characters with an empty string
# so that they are removed from the text
# the output of this function is the text without punctuation characters
```

```python
df['text'] = df['text'].apply(remove_Punc)
```

```python
def remove_numbers(text):
    new = ""
    for i in text:
        if not i.isdigit():
            new = new + i
    return new
df['text'] = df['text'].apply(remove_numbers)
# remove the numbers from the text column
```

```python
# remove the URLs $ Links from the text column emoji etc
def remove_emojis(txt):
    new = ""
    for i in txt:
        if i.isascii():
            new +=i
    return new

df['text'] = df['text'].apply(remove_emojis)
# ye remove_emojis function me txt ko input le raha hai aur usme se emojis ko remove kar raha hai
# isascii() method is used to check if the character is an ASCII character or not
# if the character is an ASCII character then it is added to the new string
# if the character is not an ASCII character then it is not added to the new string
# the output of this function is the text without emojis
```

```python
# remove the stop words from the text column

import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
nltk.download('stopwords')
nltk.download('punkt')
stop_words = set(stopwords.words('english'))
def remove_stopwords(text):
    new = ""
    for i in text.split():
        if i not in stop_words:
            new = new + " " + i
    return new
df['text'] = df['text'].apply(remove_stopwords)
# ye remove_stopwords function me text ko input le raha hai aur usme se stop words ko remove kar raha hai
# stop words are the common words that do not add much meaning to the text such as is, the, a, an etc
# stop_words is a set that contains all the stop words
# text.split() is used to split the text into words and then we are checking if each word is in the stop_words set or not
# if the word is not in the stop_words set then it is added to the new string
# the output of this function is the text without stop words
```
