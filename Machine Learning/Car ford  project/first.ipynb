```python
## we are started to make a ai model
X = df.drop(columns=['price'], axis= 1)
y = df['price']
```

```python
X
```

```python
df.columns
```

```python
X_one_enconded = pd.get_dummies(X, columns=['model', 'transmission', 'fuelType'], drop_first=True )
```

```python
X_one_enconded = X_one_enconded.astype(int)
X_one_enconded
```

```python
from sklearn.preprocessing import LabelEncoder

columns = ['model', 'transmission', 'fuelType']

Xlabel = X.copy()
label_encoder = {}

for col in columns:
    le = LabelEncoder()
    Xlabel[col] = le.fit_transform(Xlabel[col].astype(str))
    label_encoder[col] = le
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
numerical_cols = ['year', 'mileage', 'tax', 'mpg', 'engineSize']
scaler = StandardScaler()
X_one_enconded[numerical_cols] = scaler.fit_transform(X_one_enconded[numerical_cols])
```

```python
X_one_enconded
```

```python
Xlabel.columns
```

```python
Xlabel[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg', 'engineSize']] = scaler.fit_transform(Xlabel[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg', 'engineSize']])
```

```python
Xlabel
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
X_trian, X_test, y_train, y_test = train_test_split(X_one_enconded, y, test_size=0.2, random_state=42)
```

```python
model = LinearRegression()
model.fit(X_trian, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
y_test
```

```python
r2 = r2_score(y_test, y_pred)
r2
```

```python
n = X_test.shape[0]
n = X_test.shape[1]
adjusted_r2 = 1 - (1 - r2) * (n - 1) / (n - X_test.shape[1] - 1)
adjusted_r2
```
