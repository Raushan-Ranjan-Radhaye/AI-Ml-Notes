```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
## Advanced Structure : [resturent_id 2021, 2021, 2024]

sales_data = np.array([
    [1,150000, 180000, 22000, 250000], # paradise Biryani
    [2, 12000, 140000, 160000, 190000], #Beijing Bites
    [3, 20000, 210000, 240000, 300000], ##Pizza HUb
    [4, 180000, 21600, 240000, 270000],##Burger Point
    [5, 160000, 185000, 2500000, 230000], ## Chai Point
])

print("=== Zomato sales anaylises === ")
print("\n Sales data shape", sales_data.shape)
print("\n Sample data for last 3 restrunets", sales_data[0:3])
```

```python
#total sales per year
print(np.sum(sales_data, axis=1))
year_total = np.sum(sales_data, axis=0)
print(year_total)

##Minium sales Per resturunt
min_sales = np.min(sales_data[:,1], axis=0)

##Maximum sales Per resturunt
max_sales = np.max(sales_data[:,1], axis=0)
print(max_sales)

##Average sales Per resturunt
avg_sales = np.mean(sales_data[:,1:], axis=1) ## all rows se sara column lega
print(avg_sales)

cumsum = np.cumsum(sales_data[:,1:], axis=1)
print(cumsum)


plt.figure(figsize=(10,5))## te ham data ko graph me show kar rhai hai
plt.plot(np.mean( cumsum, axis=0))##
plt.title("Average cumlative sales acreoss all resturants")## graph me ye title hoag 
plt.xlabel("Year")## garphye x-axis me ye message show hoga
plt.ylabel("Sales")## y axis me ye message show hoga
plt.grid(True)## grid show hoga box ke shape me
plt.show()## show hoga
```

```python
vector1 = np.array([1,2,3,4,5,6])
vector2 = np.array([7,8,9,10,11,12])

print("Vector addition:", vector1 + vector2)
print('Multiplication vector', vector1*vector2)
print("\nDot product", np.dot(vector1, vector2))

angle = np.arccos(np.dot(vector1, vector2) / (np.linalg.norm(vector1) * np.linalg.norm(vector2)))
print("Angle between vectors", angle)
```
